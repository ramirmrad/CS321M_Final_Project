```python
from datasets import load_dataset

LETTER_CHOICES = ["A", "B", "C", "D"]


def load_gsm8k_pilot(n_items=N_GSM8K_PILOT):
    ds = load_dataset("openai/gsm8k", "main", split="test")
    ds = ds.shuffle(seed=SEED).select(range(n_items))
    items = []
    for i, ex in enumerate(ds):
        items.append({
            "id": f"gsm8k_{i}",
            "prompt": f"Solve step by step.\nQuestion: {ex['question']}\nAnswer:",
            "answer": ex["answer"].split("####")[-1].strip(),
            "format": "generative",
            "construct": "math_reasoning",
            "analysis_source": "projected",
        })
    return items


def load_mmlu_full(n_items=N_FULL_BENCHMARK):
    ds = load_dataset("cais/mmlu", "all", split="test")
    ds = ds.shuffle(seed=SEED).select(range(n_items))
    items = []
    for i, ex in enumerate(ds):
        choices = list(ex["choices"])
        answer_idx = int(ex["answer"])
        prompt = (
            f"Question: {ex['question']}\n" +
            "\n".join(f"{LETTER_CHOICES[j]}. {choice}" for j, choice in enumerate(choices)) +
            "\nAnswer:"
        )
        items.append({
            "id": f"mmlu_{i}",
            "prompt": prompt,
            "choices": choices,
            "answer": LETTER_CHOICES[answer_idx],
            "answer_idx": answer_idx,
            "format": "mcq",
            "construct": "knowledge_reasoning",
            "analysis_source": "observed_full",
        })
    return items


def load_hellaswag_full(n_items=N_FULL_BENCHMARK):
    ds = load_dataset("Rowan/hellaswag", split="validation")
    ds = ds.shuffle(seed=SEED).select(range(n_items))
    items = []
    for i, ex in enumerate(ds):
        endings = list(ex["endings"])
        answer_idx = int(ex["label"])
        prompt = (
            f"Context: {ex['ctx']}\n" +
            "\n".join(f"{LETTER_CHOICES[j]}. {ending}" for j, ending in enumerate(endings)) +
            "\nAnswer:"
        )
        items.append({
            "id": f"hellaswag_{i}",
            "prompt": prompt,
            "choices": endings,
            "answer": LETTER_CHOICES[answer_idx],
            "answer_idx": answer_idx,
            "format": "mcq",
            "construct": "commonsense_reasoning",
            "analysis_source": "observed_full",
        })
    return items


PILOT_ITEMS = {
    "gsm8k": load_gsm8k_pilot(),
    "mmlu": load_mmlu_full(),
    "hellaswag": load_hellaswag_full(),
}
BENCHMARKS = list(PILOT_ITEMS.keys())
FULL_BENCHMARKS = ["mmlu", "hellaswag"]
PROJECTED_BENCHMARKS = ["gsm8k"]
for bname, items in PILOT_ITEMS.items():
    print(f"{bname}: {len(items)} items loaded ({items[0]['analysis_source']})")
```

```python
"""---
## Cell 4 — Logit Collection (GPU required)
max_new_tokens=256 so generative completions are not truncated.
Temperature fix applied: never divide by near-zero.
"""

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from tqdm.auto import tqdm

LOGIT_CACHE_VERSION = 2  # v2 fixes cached decoding attention masks.

_tokenizer_cache = {}
def get_tokenizer(model_name):
    if model_name not in _tokenizer_cache:
        _tokenizer_cache[model_name] = AutoTokenizer.from_pretrained(
            model_name, trust_remote_code=True)
    return _tokenizer_cache[model_name]

def get_logit_cache_path(model_name, benchmark, run=0):
    safe = model_name.replace("/", "__")
    return LOGIT_DIR / f"{safe}__{benchmark}__run{run}.npz"

@torch.no_grad()
def collect_logits_pilot(model_name, items, benchmark_name,
                          max_new_tokens=256, temperature=0.7,
                          top_k_store=TOP_K_STORE, run=0):
    cache_path = get_logit_cache_path(model_name, benchmark_name, run)
    if cache_path.exists():
        try:
            cached = np.load(cache_path, allow_pickle=True)
            version_ok = ("cache_version" in cached.files and
                          int(cached["cache_version"][0]) == LOGIT_CACHE_VERSION)
            cached.close()
        except Exception:
            version_ok = False
        if version_ok:
            print(f"  Cache hit: {cache_path.name}")
            return
        print(f"  Stale cache, rebuilding: {cache_path.name}")
        cache_path.unlink()
    print(f"  Loading {model_name} ...")
    tok = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True,
                                         use_fast=True)
    model = AutoModelForCausalLM.from_pretrained(
        model_name, torch_dtype=torch.float16,
        device_map="auto", trust_remote_code=True)
    model.eval()
    rng = torch.Generator(device="cuda")
    rng.manual_seed(SEED + run)
    all_records = []
    for item in tqdm(items, desc=f"{model_name.split('/')[-1]}/{benchmark_name}"):
        inputs   = tok(item["prompt"], return_tensors="pt").to(model.device)
        past     = None
        cur_inp  = inputs["input_ids"]
        cur_attn = inputs["attention_mask"]
        gen_ids  = []
        topk_ids_all, topk_lps_all = [], []
        for step in range(max_new_tokens):
            out    = model(input_ids=cur_inp, attention_mask=cur_attn,
                           past_key_values=past, use_cache=True)
            past   = out.past_key_values
            logits = out.logits[0, -1, :]
            logits = torch.nan_to_num(logits, nan=0.0, posinf=1e4, neginf=-1e4)
            if temperature < 1e-5:
                log_probs = torch.log_softmax(logits, dim=-1)
            else:
                log_probs = torch.log_softmax(logits / temperature, dim=-1)
            log_probs = torch.nan_to_num(log_probs, nan=-1e9, posinf=0.0, neginf=-1e9)
            topk = torch.topk(log_probs, k=min(top_k_store, log_probs.shape[0]))
            topk_ids_all.append(topk.indices.cpu().numpy().astype(np.int32))
            topk_lps_all.append(topk.values.cpu().numpy().astype(np.float32))
            probs = torch.exp(log_probs).clamp(min=0.0)
            total = probs.sum()
            if total <= 0 or not torch.isfinite(total):
                next_id = torch.argmax(log_probs, keepdim=True).unsqueeze(0)
            else:
                probs   = probs / total
                next_id = torch.multinomial(probs, 1, generator=rng)
            gen_ids.append(next_id.item())
            if tok.eos_token_id is not None and next_id.item() == tok.eos_token_id:
                break
            cur_inp = next_id.unsqueeze(0)
            cur_attn = torch.cat([
                cur_attn,
                torch.ones(1, 1, device=model.device, dtype=cur_attn.dtype),
            ], dim=1)
        all_records.append({
            "item_id":       item["id"],
            "gen_ids":       np.array(gen_ids, dtype=np.int32),
            "topk_ids":      np.array(topk_ids_all, dtype=object),
            "topk_logprobs": np.array(topk_lps_all, dtype=object),
        })
    np.savez_compressed(
        cache_path,
        cache_version = np.array([LOGIT_CACHE_VERSION], dtype=np.int32),
        item_ids      = np.array([r["item_id"]       for r in all_records]),
        gen_ids       = np.array([r["gen_ids"]        for r in all_records], dtype=object),
        topk_ids      = np.array([r["topk_ids"]       for r in all_records], dtype=object),
        topk_logprobs = np.array([r["topk_logprobs"]  for r in all_records], dtype=object),
    )
    print(f"  Saved → {cache_path}")
    del model; del tok
    gc.collect(); torch.cuda.empty_cache()

# This notebook is now analysis-first: it loads existing cache/logits/*.npz files.
# Set RUN_LOGIT_COLLECTION=True only if you intentionally want to create missing caches.
RUN_LOGIT_COLLECTION = False

if RUN_LOGIT_COLLECTION:
    os.environ["CUDA_LAUNCH_BLOCKING"] = "1"
    MAX_NEW_TOKENS = {"gsm8k": 384, "mmlu": 1, "hellaswag": 1}
    for model_name in ALL_MODELS:
        for bname, items in PILOT_ITEMS.items():
            try:
                collect_logits_pilot(
                    model_name, items, bname,
                    max_new_tokens=MAX_NEW_TOKENS.get(bname, 1),
                )
            except Exception as e:
                print(f"ERROR {model_name}/{bname}: {e}")
                gc.collect(); torch.cuda.empty_cache()
else:
    print("Skipping logit collection; using existing cache/logits/*.npz files.")
```

```python
"""---
## Cell 5 — Cache Loader + Speculative Decoding Simulator
"""

from transformers import AutoTokenizer
VOCAB_SIZES = {}
for model_name in ALL_MODELS:
    tok = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
    VOCAB_SIZES[model_name] = len(tok)

def load_logit_cache(model_name, benchmark, run=0):
    path = get_logit_cache_path(model_name, benchmark, run)
    data = np.load(path, allow_pickle=True)
    # Older one-token MCQ caches do not have cache_version, but the old
    # attention-mask bug only corrupted multi-token generative caches.
    if "cache_version" in data.files:
        version_ok = int(data["cache_version"][0]) == LOGIT_CACHE_VERSION
    else:
        version_ok = benchmark in FULL_BENCHMARKS
    if not version_ok:
        data.close()
        raise ValueError(
            f"Stale logit cache: {path.name}. Re-run collection to rebuild it."
        )
    records = {}
    for i, iid in enumerate(data["item_ids"]):
        records[iid] = {
            "gen_ids":       data["gen_ids"][i],
            "topk_ids":      data["topk_ids"][i],
            "topk_logprobs": data["topk_logprobs"][i],
        }
    data.close()
    return records

def topk_to_full_dist(topk_ids, topk_logprobs, vocab_size,
                       temperature=None, collection_temp=COLLECTION_TEMP):
    """
    Reconstruct a sparse probability array from top-k storage.

    Parameters
    ----------
    temperature : float or None
        If None, use the stored logprobs as-is (original behavior).
        If float, rescale the logprobs to simulate a different temperature.
    collection_temp : float
        The temperature used when the logprobs were originally collected.
        Only used when temperature is not None.
    """
    probs = np.zeros(vocab_size, dtype=np.float32)
    lp    = topk_logprobs.astype(np.float64)
    lp    = np.nan_to_num(lp, nan=-1e9, posinf=-1e9, neginf=-1e9)

    if temperature is not None and abs(temperature - collection_temp) > 1e-6:
        # Undo collection temperature to recover approximate logit scale,
        # then apply new temperature.
        # log_softmax output satisfies: lp ≈ logit/T - log Z
        # Rescaling: lp_new ≈ lp * (collection_temp / temperature)
        # This is exact when all vocab mass is in the top-k; approximate otherwise.
        scale = collection_temp / temperature
        lp    = lp * scale

    lp    -= lp.max()   # numerical stability
    p_topk = np.exp(lp)
    p_topk = np.clip(p_topk, 0.0, None)
    total  = p_topk.sum()
    if total <= 0 or not np.isfinite(total):
        p_topk    = np.zeros_like(p_topk)
        p_topk[0] = 1.0
    else:
        p_topk /= total
    probs[topk_ids.astype(np.int32)] = p_topk
    return probs


def simulate_speculative_step(p_target, p_draft, proposed_token,
                               epsilon=0.0, rng=None):
    if rng is None:
        rng = np.random.default_rng(SEED)
    q = p_draft[proposed_token]; p = p_target[proposed_token]
    if epsilon == 0.0:
        accept_prob = min(1.0, p / (q + 1e-12))
        accepted    = rng.random() < accept_prob
    else:
        ratio    = p / (q + 1e-12)
        accepted = ratio >= (1.0 - epsilon)
        accept_prob = float(accepted)
    if accepted:
        return proposed_token, True, accept_prob
    else:
        if epsilon == 0.0:
            residual = np.maximum(0.0, p_target - p_draft)
            s = residual.sum()
            corrected = (rng.choice(len(residual), p=residual/s)
                         if s > 0 else np.argmax(p_target))
        else:
            corrected = rng.choice(len(p_target), p=p_target/p_target.sum())
        return corrected, False, accept_prob

def simulate_condition(draft_cache, target_cache, items, condition_key,
                        epsilon, run_seed, vocab_size, temperature=None):
    """
    Simulate one decoding condition for all items.

    Parameters
    ----------
    epsilon     : float or None — speculative threshold (None = no spec decoding)
    temperature : float or None — replay temperature (None = use stored probs as-is)
    """
    rng     = np.random.default_rng(run_seed)
    results = {}

    is_temp_condition = (epsilon is None and temperature is not None)

    for item in items:
        iid = item["id"]
        d   = draft_cache.get(iid)
        t   = target_cache.get(iid)
        if d is None or t is None:
            continue

        n_steps = len(t["topk_ids"])
        tokens, accepts, accept_probs = [], [], []

        for step in range(n_steps):
            t_ids = t["topk_ids"][step]
            t_lp  = t["topk_logprobs"][step]
            d_ids = d["topk_ids"][step]      if step < len(d["topk_ids"])      else t_ids
            d_lp  = d["topk_logprobs"][step] if step < len(d["topk_logprobs"]) else t_lp

            # Build target distribution — apply replay temperature if set
            p_target = topk_to_full_dist(t_ids, t_lp, vocab_size,
                                          temperature=temperature)
            p_draft  = topk_to_full_dist(d_ids, d_lp, vocab_size,
                                          temperature=None)  # draft always at collection temp

            if condition_key == "A" or is_temp_condition:
                # Baseline or temperature replay: sample directly from target
                # (temperature is already baked into p_target above)
                if temperature is None or condition_key == "A":
                    # Condition A: use the exact stored sampled token (deterministic replay)
                    chosen = (t["gen_ids"][step] if step < len(t["gen_ids"])
                              else np.argmax(p_target))
                    tokens.append(int(chosen))
                else:
                    # Temperature condition: resample from rescaled distribution
                    p_safe = np.clip(p_target, 0.0, None)
                    total  = p_safe.sum()
                    if total > 0:
                        p_safe /= total
                        chosen = rng.choice(len(p_safe), p=p_safe)
                    else:
                        chosen = np.argmax(p_target)
                    tokens.append(int(chosen))
                accepts.append(True)
                accept_probs.append(1.0)

            else:
                # Speculative decoding conditions (B, C_*)
                proposed = (d["gen_ids"][step] if step < len(d["gen_ids"])
                            else np.argmax(p_draft))
                tok_out, acc, ap = simulate_speculative_step(
                    p_target, p_draft, int(proposed),
                    epsilon=epsilon, rng=rng)
                tokens.append(tok_out)
                accepts.append(acc)
                accept_probs.append(ap)

        results[iid] = {
            "tokens":       tokens,
            "accept_flags": accepts,
            "accept_rate":  float(np.mean(accepts)) if accepts else 1.0,
        }

    return results

print("Simulator defined.")
```

```python
"""---
## Cell 6 — Scoring
HumanEval is intentionally removed from the main project analysis.
"""


def extract_number(text):
    text = text.replace(",", "")
    if "####" in text:
        text = text.split("####")[-1]

    patterns = [
        r"(?:answer is|answer:|therefore|so|will be)\s*\$?\s*(-?\d+(?:\.\d+)?)",
        r"=\s*\$?\s*(-?\d+(?:\.\d+)?)\s*(?:years?|dollars?|$|\.)",
    ]
    for pat in patterns:
        matches = re.findall(pat, text, flags=re.IGNORECASE)
        if matches:
            return matches[-1]

    nums = re.findall(r"(?<![\w.])-?\d+(?:\.\d+)?(?![\w.])", text)
    return nums[-1] if nums else None


def extract_mcq_choice(text):
    text = text.strip()
    if not text:
        return None
    match = re.search(r"\b([ABCD])\b", text.upper())
    if match:
        return match.group(1)
    match = re.search(r"\b([0-3])\b", text)
    if match:
        return LETTER_CHOICES[int(match.group(1))]
    return None


def score_gsm8k(item, sim_result, tokenizer):
    if not sim_result["tokens"]:
        return 0
    generated = tokenizer.decode(sim_result["tokens"]).strip()
    pred = extract_number(generated)
    gold = extract_number(item["answer"])
    return int(pred is not None and gold is not None
               and abs(float(pred) - float(gold)) < 1e-3)


def score_mcq(item, sim_result, tokenizer):
    if not sim_result["tokens"]:
        return 0
    generated = tokenizer.decode(sim_result["tokens"]).strip()
    pred = extract_mcq_choice(generated)
    return int(pred == item["answer"])


def score_item(item, sim_result, tokenizer, benchmark_name):
    if benchmark_name == "gsm8k":
        return score_gsm8k(item, sim_result, tokenizer)
    if benchmark_name in {"mmlu", "hellaswag"}:
        return score_mcq(item, sim_result, tokenizer)
    return 0

print("Scoring functions defined.")
```

```python
def use_cached_generation(target_cache, draft_cache, items,
                           condition_key, epsilon, run_seed,
                           vocab_size, temperature=None):
    """
    Use stored gen_ids directly instead of doing any extra model generation.

    Condition A: use target gen_ids (original target generation).
    Condition B: exact speculation is measurement-equivalent to the target
                 output, so keep target gen_ids while estimating acceptance.
    Approximate speculation and temperature controls are handled by
    simulate_condition(), which replays from stored top-k logits only.
    """
    rng     = np.random.default_rng(run_seed)
    results = {}

    for item in items:
        iid = item["id"]
        d   = draft_cache.get(iid)
        t   = target_cache.get(iid)
        if d is None or t is None:
            continue

        # For all conditions, start from the stored target generation
        tokens = list(t["gen_ids"])

        if condition_key == "A" or temperature is not None:
            # Baseline or temperature: use target tokens as-is
            accept_rate = 1.0

        else:
            # Speculative conditions: simulate acceptance on stored tokens
            # This gives a valid accept_rate without resampling distributions
            n_steps = min(len(t["topk_ids"]), len(t["gen_ids"]),
                          len(d["gen_ids"]))
            accepts = []
            for step in range(n_steps):
                t_ids = t["topk_ids"][step]
                t_lp  = t["topk_logprobs"][step]
                d_ids = d["topk_ids"][step]
                d_lp  = d["topk_logprobs"][step]
                p_target = topk_to_full_dist(t_ids, t_lp, vocab_size,
                                              temperature=temperature)
                p_draft  = topk_to_full_dist(d_ids, d_lp, vocab_size)
                proposed = int(d["gen_ids"][step])
                _, acc, _ = simulate_speculative_step(
                    p_target, p_draft, proposed,
                    epsilon=epsilon if epsilon is not None else 0.0,
                    rng=rng)
                accepts.append(acc)
            accept_rate = float(np.mean(accepts)) if accepts else 1.0

        results[iid] = {
            "tokens":       tokens,
            "accept_flags": [],
            "accept_rate":  accept_rate,
        }

    return results
```

```python
rows = []

# ── Standalone draft accuracy (needed for Edit 6 moderator analysis) ─────────
# Score each draft model independently using Condition A logic (no spec decoding).
# We reuse the existing target_cache infrastructure — just treat draft as target.
draft_standalone_rows = []

N_RUNS = 1

for pair in MODEL_PAIRS:
    draft_name  = pair["draft"]
    target_name = pair["target"]
    vocab_size  = VOCAB_SIZES[target_name]
    pair_label  = f"{draft_name.split('/')[-1]}→{target_name.split('/')[-1]}"
    tokenizer   = get_tokenizer(target_name)
    draft_tokenizer = get_tokenizer(draft_name)

    for bname, items in PILOT_ITEMS.items():
        for run in tqdm(range(N_RUNS), desc=f"{pair_label}/{bname}"):
            try:
                draft_cache  = load_logit_cache(draft_name,  bname, run)
                target_cache = load_logit_cache(target_name, bname, run)
            except FileNotFoundError as e:
                print(f"Skipping {pair_label}/{bname}: missing cache ({e})")
                continue
            except ValueError as e:
                print(f"Skipping {pair_label}/{bname}: {e}")
                continue

            # ── Standalone draft scoring ──────────────────────────────────
            # Use draft cache as both draft and target (i.e. score draft alone)
            draft_standalone = use_cached_generation(
                draft_cache, draft_cache, items,
                condition_key = "A",
                epsilon       = None,
                run_seed      = SEED + run * 1000,
                vocab_size    = vocab_size,
                temperature   = None,
            )
            # draft_standalone = enrich_sim_results_with_topk(
            #     draft_standalone, draft_cache, items)
            for item in items:
                iid = item["id"]
                if iid not in draft_standalone: continue
                correct = score_item(item, draft_standalone[iid], draft_tokenizer, bname)
                draft_standalone_rows.append({
                    "model_pair": pair_label,
                    "draft_model": draft_name,
                    "benchmark":  bname,
                    "run":        run,
                    "item_id":    iid,
                    "correct":    correct,
                    "construct":  item["construct"],
                    "analysis_source": item["analysis_source"],
                })

            # ── Main condition loop ───────────────────────────────────────
            for cond_key, cond_cfg in CONDITIONS.items():
                epsilon     = cond_cfg["epsilon"]
                temperature = cond_cfg["temperature"]

                if cond_key in ["A", "B"]:
                    # Baseline and exact speculation should decode to the same target output.
                    # For B we still estimate token-level acceptance from stored logits.
                    sim_results = use_cached_generation(
                        target_cache, draft_cache, items,
                        condition_key = cond_key,
                        epsilon       = epsilon,
                        run_seed      = SEED + run * 1000,
                        vocab_size    = vocab_size,
                        temperature   = None,
                    )
                else:
                    # Approximate speculation and temperature controls are offline replays
                    # from stored top-k logits; no extra model generation is performed.
                    sim_results = simulate_condition(
                        draft_cache, target_cache, items,
                        condition_key = cond_key,
                        epsilon       = epsilon,
                        run_seed      = SEED + run * 1000,
                        vocab_size    = vocab_size,
                        temperature   = temperature,
                    )
                # sim_results = enrich_sim_results_with_topk(
                #     sim_results, target_cache, items)

                for item in items:
                    iid = item["id"]
                    if iid not in sim_results: continue
                    sr      = sim_results[iid]
                    correct = score_item(item, sr, tokenizer, bname)
                    rows.append({
                        "model_pair":  pair_label,
                        "benchmark":   bname,
                        "condition":   cond_key,
                        "cond_type":   ("spec" if cond_key in SPEC_CONDITIONS
                                        else "temp"),
                        "run":         run,
                        "item_id":     iid,
                        "correct":     correct,
                        "token_ids":   " ".join(map(str, sr["tokens"])),
                        "n_tokens":    len(sr["tokens"]),
                        "accept_rate": sr["accept_rate"],
                        "construct":   item["construct"],
                        "analysis_source": item["analysis_source"],
                        "temperature": temperature if temperature is not None
                                       else COLLECTION_TEMP,
                    })

df = pd.DataFrame(rows)
pilot_df = df.copy()
draft_df = pd.DataFrame(draft_standalone_rows)

# ── Draft accuracy gap table ──────────────────────────────────────────────────
# accuracy_gap = target_acc_A - draft_standalone_acc
# This is the key moderator variable for Edit 6.
target_acc = (df[df["condition"]=="A"]
               .groupby(["model_pair","benchmark","run"])
               ["correct"].mean().reset_index(name="target_acc"))

draft_acc  = (draft_df.groupby(["model_pair","benchmark","run"])
                      ["correct"].mean().reset_index(name="draft_acc"))

acc_gap_df = target_acc.merge(draft_acc, on=["model_pair","benchmark","run"])
acc_gap_df["accuracy_gap"] = acc_gap_df["target_acc"] - acc_gap_df["draft_acc"]

print("\n=== Draft vs Target Accuracy (moderator variable) ===")
print(acc_gap_df.groupby(["model_pair","benchmark"])
                [["target_acc","draft_acc","accuracy_gap"]]
                .mean().round(4).to_string())

df.to_csv(RESULT_DIR / "raw_results.csv", index=False)
draft_df.to_csv(RESULT_DIR / "draft_standalone.csv", index=False)
acc_gap_df.to_csv(RESULT_DIR / "accuracy_gaps.csv", index=False)
print(f"\nDone. {len(df):,} rows.")
```

```python
"""---
## Cell 7b — Measurement Validity Diagnostics for Speculative Decoding

Adds analyses aligned with the core project question:
1. Does speculative decoding change item outcomes?
2. Do lower acceptance rates / larger draft-target divergences predict score flips?
3. How large are speculative effects relative to temperature perturbations?

Designed to scale to later MMLU/HellaSwag additions: it groups over whatever
benchmarks are present in df and PILOT_ITEMS.
"""

MECH_MAX_STEPS = 128  # cap long generative traces so 500-item runs stay manageable
EPS = 1e-12


def _condition_alpha(condition):
    """Return the approximate acceptance threshold encoded by condition labels."""
    if condition == "B":
        return 1.0
    if condition.startswith("C_"):
        return int(condition.split("_")[1]) / 100.0
    return np.nan


def _topk_probs(topk_ids, topk_logprobs, temperature=None,
                collection_temp=COLLECTION_TEMP):
    ids = np.asarray(topk_ids, dtype=np.int64)
    lp = np.asarray(topk_logprobs, dtype=np.float64)
    lp = np.nan_to_num(lp, nan=-1e9, posinf=-1e9, neginf=-1e9)
    if temperature is not None and abs(temperature - collection_temp) > 1e-6:
        lp = lp * (collection_temp / temperature)
    lp = lp - lp.max()
    p = np.exp(lp)
    total = p.sum()
    if total <= 0 or not np.isfinite(total):
        p = np.ones_like(p) / max(1, len(p))
    else:
        p = p / total
    return ids, p


def _entropy(p):
    p = np.asarray(p, dtype=np.float64)
    p = p[p > 0]
    return float(-(p * np.log(p + EPS)).sum()) if len(p) else np.nan


def _union_distribution(ids, probs, union_ids):
    out = np.zeros(len(union_ids), dtype=np.float64)
    pos = {int(tok): j for j, tok in enumerate(union_ids)}
    for tok, prob in zip(ids, probs):
        out[pos[int(tok)]] += float(prob)
    s = out.sum()
    return out / s if s > 0 else out


def _step_mechanism_metrics(t_ids, t_lp, d_ids, d_lp, proposed_token,
                            temperature=None):
    t_ids, p_t = _topk_probs(t_ids, t_lp, temperature=temperature)
    d_ids, p_d = _topk_probs(d_ids, d_lp)
    union_ids = np.array(sorted(set(map(int, t_ids)).union(map(int, d_ids))),
                         dtype=np.int64)
    u_t = _union_distribution(t_ids, p_t, union_ids)
    u_d = _union_distribution(d_ids, p_d, union_ids)
    mix = 0.5 * (u_t + u_d)

    kl_t_m = float(np.sum(u_t * (np.log(u_t + EPS) - np.log(mix + EPS))))
    kl_d_m = float(np.sum(u_d * (np.log(u_d + EPS) - np.log(mix + EPS))))
    js_div = 0.5 * (kl_t_m + kl_d_m)

    proposed_token = int(proposed_token)
    p_target_prop = float(p_t[t_ids == proposed_token].sum())
    p_draft_prop = float(p_d[d_ids == proposed_token].sum())
    accept_prob_exact = min(1.0, p_target_prop / (p_draft_prop + EPS))

    return {
        "target_entropy_topk": _entropy(p_t),
        "draft_entropy_topk": _entropy(p_d),
        "js_div_topk": js_div,
        "target_prob_draft_token": p_target_prop,
        "draft_prob_draft_token": p_draft_prop,
        "exact_accept_prob": accept_prob_exact,
        "exact_reject_prob": 1.0 - accept_prob_exact,
    }


def build_mechanism_diagnostics(model_pairs=MODEL_PAIRS, pilot_items=PILOT_ITEMS,
                                run=0, max_steps=MECH_MAX_STEPS):
    rows = []
    for pair in model_pairs:
        draft_name = pair["draft"]
        target_name = pair["target"]
        pair_label = f"{draft_name.split('/')[-1]}→{target_name.split('/')[-1]}"
        for bname, items in pilot_items.items():
            try:
                draft_cache = load_logit_cache(draft_name, bname, run)
                target_cache = load_logit_cache(target_name, bname, run)
            except Exception as e:
                print(f"Skipping mechanism diagnostics for {pair_label}/{bname}: {e}")
                continue

            for item in items:
                iid = item["id"]
                d = draft_cache.get(iid)
                t = target_cache.get(iid)
                if d is None or t is None:
                    continue
                n_steps = min(len(t["topk_ids"]), len(d["topk_ids"]),
                              len(d["gen_ids"]), max_steps)
                if n_steps == 0:
                    continue
                step_rows = []
                for step in range(n_steps):
                    step_rows.append(_step_mechanism_metrics(
                        t["topk_ids"][step], t["topk_logprobs"][step],
                        d["topk_ids"][step], d["topk_logprobs"][step],
                        d["gen_ids"][step],
                    ))
                step_df = pd.DataFrame(step_rows)
                rows.append({
                    "model_pair": pair_label,
                    "benchmark": bname,
                    "item_id": iid,
                    "construct": item.get("construct", bname),
                    "n_diag_steps": n_steps,
                    "mean_target_entropy_topk": step_df["target_entropy_topk"].mean(),
                    "mean_draft_entropy_topk": step_df["draft_entropy_topk"].mean(),
                    "mean_js_div_topk": step_df["js_div_topk"].mean(),
                    "mean_target_prob_draft_token": step_df["target_prob_draft_token"].mean(),
                    "mean_exact_accept_prob": step_df["exact_accept_prob"].mean(),
                    "mean_exact_reject_prob": step_df["exact_reject_prob"].mean(),
                })
    return pd.DataFrame(rows)


def build_score_flip_df(df_in):
    rows = []
    for (pair, bname), grp in df_in.groupby(["model_pair", "benchmark"]):
        base = grp[grp["condition"] == "A"].set_index("item_id")
        if base.empty:
            continue
        for cond, comp in grp[grp["condition"] != "A"].groupby("condition"):
            comp = comp.set_index("item_id")
            shared = base.index.intersection(comp.index)
            for iid in shared:
                base_tokens = str(base.loc[iid, "token_ids"]) if "token_ids" in base else ""
                comp_tokens = str(comp.loc[iid, "token_ids"]) if "token_ids" in comp else ""
                token_changed = int(base_tokens != comp_tokens)
                rows.append({
                    "model_pair": pair,
                    "benchmark": bname,
                    "condition": cond,
                    "cond_type": comp.loc[iid, "cond_type"],
                    "item_id": iid,
                    "source_item_id": comp.loc[iid, "source_item_id"] if "source_item_id" in comp else iid,
                    "baseline_correct": int(base.loc[iid, "correct"]),
                    "condition_correct": int(comp.loc[iid, "correct"]),
                    "score_flip": int(base.loc[iid, "correct"] != comp.loc[iid, "correct"]),
                    "token_changed": token_changed,
                    "token_change_no_score_flip": int(token_changed and base.loc[iid, "correct"] == comp.loc[iid, "correct"]),
                    "accept_rate": float(comp.loc[iid, "accept_rate"]),
                    "temperature": float(comp.loc[iid, "temperature"]),
                    "construct": comp.loc[iid, "construct"],
                })
    return pd.DataFrame(rows)


def build_condition_impact_df(df_in, score_flips):
    acc = (df_in.groupby(["model_pair", "benchmark", "condition", "cond_type",
                          "temperature", "run"])
                 .agg(accuracy=("correct", "mean"),
                      mean_accept_rate=("accept_rate", "mean"),
                      n_items=("item_id", "nunique"))
                 .reset_index())
    base = (acc[acc["condition"] == "A"]
              [["model_pair", "benchmark", "run", "accuracy"]]
              .rename(columns={"accuracy": "baseline_accuracy"}))
    out = acc.merge(base, on=["model_pair", "benchmark", "run"], how="left")
    out["accuracy_delta_vs_A"] = out["accuracy"] - out["baseline_accuracy"]

    if len(score_flips):
        flips = (score_flips.groupby(["model_pair", "benchmark", "condition"])
                            .agg(score_flip_rate=("score_flip", "mean"),
                                 n_flips=("score_flip", "sum"),
                                 token_change_rate=("token_changed", "mean"),
                                 token_change_no_score_flip_rate=("token_change_no_score_flip", "mean"))
                            .reset_index())
        out = out.merge(flips, on=["model_pair", "benchmark", "condition"], how="left")
    else:
        out["score_flip_rate"] = np.nan
        out["n_flips"] = np.nan
        out["token_change_rate"] = np.nan
        out["token_change_no_score_flip_rate"] = np.nan
    out["alpha_threshold"] = out["condition"].map(_condition_alpha)
    return out


# This cell only defines validity-diagnostic helpers. The diagnostics are run
# after analysis_df is built so GSM8K uses its paired n=200 projection and
# MMLU/HellaSwag use their full observed 500-item caches.
```

```python
"""---
## Cell 8 — Build Final Analysis Dataset

Final analysis uses:
- MMLU: full observed 500 cached items
- HellaSwag: full observed 500 cached items
- GSM8K: paired bootstrap projection from the 20-item pilot to n=200

No downstream analysis is run on the raw 20-item GSM8K pilot.
"""


def one_over_sqrt_n(n, a, b):
    return a / np.sqrt(n) + b


def bootstrap_extrapolation(correct_vec, virtual_ns=VIRTUAL_NS,
                            n_boot=N_BOOT, seed=SEED):
    rng = np.random.default_rng(seed)
    x = np.array(correct_vec, dtype=float)
    obs_mean = x.mean()
    n_obs = len(x)
    means, ci_lo, ci_hi = [], [], []
    for n_virt in virtual_ns:
        n_sample = min(n_virt, n_obs)
        boot_means = np.array([
            rng.choice(x, size=n_sample, replace=True).mean()
            for _ in range(n_boot)
        ])
        means.append(boot_means.mean())
        ci_lo.append(np.percentile(boot_means, 2.5))
        ci_hi.append(np.percentile(boot_means, 97.5))
    means = np.array(means)
    ci_lo = np.array(ci_lo)
    ci_hi = np.array(ci_hi)
    half_width = (ci_hi - ci_lo) / 2.0
    observed_mask = np.array(virtual_ns) <= n_obs
    ns_arr = np.array(virtual_ns, dtype=float)
    try:
        popt, _ = curve_fit(
            one_over_sqrt_n,
            ns_arr[observed_mask], half_width[observed_mask],
            p0=[0.5, 0.01], maxfev=5000,
            bounds=([0, 0], [10, 1]),
        )
        fitted_a, fitted_b = popt
    except RuntimeError:
        fitted_a = x.std() * 1.96
        fitted_b = 0.0
    projected_half = one_over_sqrt_n(N_GSM8K_PROJECTED, fitted_a, fitted_b)
    return {
        "ns": virtual_ns,
        "obs_mean": obs_mean,
        "n_obs": n_obs,
        "boot_means": means,
        "boot_ci_lo": ci_lo,
        "boot_ci_hi": ci_hi,
        "half_width": half_width,
        "fitted_a": fitted_a,
        "fitted_b": fitted_b,
        "projected_mean": obs_mean,
        "projected_ci_lo": np.clip(obs_mean - projected_half, 0, 1),
        "projected_ci_hi": np.clip(obs_mean + projected_half, 0, 1),
        "projected_half": projected_half,
    }


observed_df = df.copy()
observed_df["source_item_id"] = observed_df["item_id"]

# Keep full cached benchmarks as observed data.
full_observed_df = observed_df[observed_df["benchmark"].isin(FULL_BENCHMARKS)].copy()
full_observed_df["analysis_source"] = "observed_full_500"

# Project GSM8K in a paired way: sample source item IDs once, then copy every
# condition for that source item under the same projected item_id.
gsm_proj_rows = []
rng_proj = np.random.default_rng(SEED + 999)
gsm_obs = observed_df[observed_df["benchmark"] == "gsm8k"].copy()

for (pair, run), grp in gsm_obs.groupby(["model_pair", "run"]):
    item_ids = np.array(sorted(grp["item_id"].unique()))
    if len(item_ids) == 0:
        continue
    sampled_ids = rng_proj.choice(item_ids, size=N_GSM8K_PROJECTED, replace=True)
    by_item = {iid: item_grp.copy() for iid, item_grp in grp.groupby("item_id")}
    for j, source_iid in enumerate(sampled_ids):
        projected_iid = f"gsm8k_proj_{j}"
        item_rows = by_item[source_iid].copy()
        item_rows["source_item_id"] = source_iid
        item_rows["item_id"] = projected_iid
        item_rows["analysis_source"] = "projected_gsm8k_200"
        gsm_proj_rows.extend(item_rows.to_dict("records"))

gsm_projected_df = pd.DataFrame(gsm_proj_rows)
analysis_df = pd.concat([full_observed_df, gsm_projected_df], ignore_index=True)

# Backward-compatible aliases for older analysis cells, now both point at the
# final mixed dataset. There is no raw-20 GSM8K analysis below this cell.
pilot_df = analysis_df.copy()
projected_df = analysis_df.copy()

# Extrapolation summaries are computed only from the observed GSM8K pilot.
extrap_results = {}
for (pair, cond), grp in gsm_obs.groupby(["model_pair", "condition"]):
    extrap_results[(pair, "gsm8k", cond)] = bootstrap_extrapolation(grp["correct"].values)

analysis_df.to_csv(RESULT_DIR / "analysis_dataset.csv", index=False)
gsm_projected_df.to_csv(RESULT_DIR / "gsm8k_projected_200.csv", index=False)
full_observed_df.to_csv(RESULT_DIR / "full_observed_mmlu_hellaswag.csv", index=False)

print("\nFinal analysis dataset:")
print(analysis_df.groupby(["benchmark", "analysis_source"])["item_id"]
                 .nunique().to_string())
print(f"\nRows in analysis_df: {len(analysis_df):,}")
```

```python
"""---
## Cell 9 — Final Measurement Validity Diagnostics
Runs only on analysis_df: full MMLU/HellaSwag plus projected GSM8K.
"""

mechanism_df = build_mechanism_diagnostics()
score_flip_df = build_score_flip_df(analysis_df)
condition_impact_df = build_condition_impact_df(analysis_df, score_flip_df)
spec_curve_df = condition_impact_df[
    condition_impact_df["condition"].isin([c for c in SPEC_CONDITIONS if c != "A"])
].copy()

# Mechanism diagnostics are measured on source cached items. For projected GSM8K,
# join via source_item_id; for full MCQ benchmarks, source_item_id == item_id.
if len(score_flip_df) and len(mechanism_df):
    flip_join = score_flip_df.copy()
    if "source_item_id" not in flip_join.columns:
        source_map = analysis_df[["model_pair", "benchmark", "item_id", "source_item_id"]].drop_duplicates()
        flip_join = flip_join.merge(source_map, on=["model_pair", "benchmark", "item_id"], how="left")
    flip_join["mechanism_item_id"] = flip_join["source_item_id"].fillna(flip_join["item_id"])
    flip_mechanism_df = flip_join.merge(
        mechanism_df.rename(columns={"item_id": "mechanism_item_id"}),
        on=["model_pair", "benchmark", "mechanism_item_id", "construct"],
        how="left",
    )
else:
    flip_mechanism_df = pd.DataFrame()

assoc_rows = []
predictors = [
    "accept_rate",
    "mean_js_div_topk",
    "mean_target_entropy_topk",
    "mean_exact_reject_prob",
    "mean_target_prob_draft_token",
]
if len(flip_mechanism_df):
    for (bname, cond), grp in flip_mechanism_df.groupby(["benchmark", "condition"]):
        if grp["score_flip"].nunique() < 2:
            continue
        for pred in predictors:
            vals = grp[pred].astype(float)
            valid = vals.notna()
            if valid.sum() < 3 or vals[valid].nunique() < 2:
                continue
            r, p = stats.pointbiserialr(grp.loc[valid, "score_flip"], vals[valid])
            assoc_rows.append({
                "benchmark": bname,
                "condition": cond,
                "predictor": pred,
                "point_biserial_r": r,
                "p": p,
                "n": int(valid.sum()),
            })
mechanism_assoc_df = pd.DataFrame(assoc_rows)

spec_max = (condition_impact_df[condition_impact_df["cond_type"] == "spec"]
              .query("condition != 'A'")
              .groupby(["model_pair", "benchmark"])
              .agg(max_abs_spec_delta=("accuracy_delta_vs_A", lambda x: np.max(np.abs(x))),
                   max_spec_flip_rate=("score_flip_rate", "max"))
              .reset_index())
temp_max = (condition_impact_df[condition_impact_df["cond_type"] == "temp"]
              .groupby(["model_pair", "benchmark"])
              .agg(max_abs_temp_delta=("accuracy_delta_vs_A", lambda x: np.max(np.abs(x))),
                   max_temp_flip_rate=("score_flip_rate", "max"))
              .reset_index())
temp_positive_control_df = spec_max.merge(temp_max, on=["model_pair", "benchmark"], how="outer")
temp_positive_control_df["spec_delta_vs_temp_delta_ratio"] = (
    temp_positive_control_df["max_abs_spec_delta"] /
    (temp_positive_control_df["max_abs_temp_delta"] + EPS)
)

print("\n=== Final Speculative Decoding Measurement Impact ===")
cols = ["model_pair", "benchmark", "condition", "accuracy_delta_vs_A",
        "score_flip_rate", "token_change_rate", "mean_accept_rate",
        "alpha_threshold", "n_items"]
print(spec_curve_df[cols].to_string(index=False, float_format="{:.4f}".format))

print("\n=== Speculation vs Temperature Positive Control ===")
print(temp_positive_control_df.to_string(index=False, float_format="{:.4f}".format))

if len(mechanism_assoc_df):
    print("\n=== Mechanism Predictors of Item Score Flips ===")
    print(mechanism_assoc_df.sort_values(["benchmark", "condition", "p"])
                            .to_string(index=False, float_format="{:.4f}".format))
else:
    print("\nNo score-flip mechanism associations estimable yet; need observed flips.")

mechanism_df.to_csv(RESULT_DIR / "mechanism_diagnostics.csv", index=False)
score_flip_df.to_csv(RESULT_DIR / "score_flips.csv", index=False)
condition_impact_df.to_csv(RESULT_DIR / "condition_impact.csv", index=False)
spec_curve_df.to_csv(RESULT_DIR / "spec_threshold_curve.csv", index=False)
flip_mechanism_df.to_csv(RESULT_DIR / "score_flip_mechanisms.csv", index=False)
mechanism_assoc_df.to_csv(RESULT_DIR / "mechanism_flip_associations.csv", index=False)
temp_positive_control_df.to_csv(RESULT_DIR / "spec_vs_temperature_positive_control.csv", index=False)

if len(spec_curve_df):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharex=True)
    plot_df = spec_curve_df.dropna(subset=["alpha_threshold"]).copy()
    sns.lineplot(data=plot_df, x="alpha_threshold", y="accuracy_delta_vs_A",
                 hue="benchmark", style="model_pair", marker="o", ax=axes[0])
    axes[0].axhline(0, color="black", lw=0.8, ls="--")
    axes[0].invert_xaxis()
    axes[0].set_title("Accuracy Change vs Approximation Threshold")
    axes[0].set_xlabel("Minimum acceptance ratio alpha")
    axes[0].set_ylabel("Accuracy delta vs baseline A")

    sns.lineplot(data=plot_df, x="alpha_threshold", y="score_flip_rate",
                 hue="benchmark", style="model_pair", marker="o", ax=axes[1])
    sns.lineplot(data=plot_df, x="alpha_threshold", y="token_change_rate",
                 hue="benchmark", style="model_pair", marker="x", legend=False,
                 alpha=0.45, ax=axes[1])
    axes[1].invert_xaxis()
    axes[1].set_title("Outcome Flips and Token Perturbations")
    axes[1].set_xlabel("Minimum acceptance ratio alpha")
    axes[1].set_ylabel("Rate vs baseline A")
    plt.tight_layout()
    plt.savefig(RESULT_DIR / "fig_spec_threshold_validity.pdf", bbox_inches="tight")
    plt.show()

if len(temp_positive_control_df):
    pc_long = temp_positive_control_df.melt(
        id_vars=["model_pair", "benchmark"],
        value_vars=["max_abs_spec_delta", "max_abs_temp_delta"],
        var_name="perturbation", value_name="max_abs_accuracy_delta",
    )
    pc_long["perturbation"] = pc_long["perturbation"].map({
        "max_abs_spec_delta": "Speculation",
        "max_abs_temp_delta": "Temperature",
    })
    plt.figure(figsize=(10, 4))
    sns.barplot(data=pc_long, x="benchmark", y="max_abs_accuracy_delta",
                hue="perturbation")
    plt.title("Speculative Decoding Effect Size vs Temperature Positive Control")
    plt.ylabel("Max |accuracy delta vs A|")
    plt.xlabel("Benchmark")
    plt.tight_layout()
    plt.savefig(RESULT_DIR / "fig_spec_vs_temperature_positive_control.pdf",
                bbox_inches="tight")
    plt.show()

if len(mechanism_assoc_df):
    plt.figure(figsize=(10, 5))
    assoc_plot = mechanism_assoc_df.copy()
    assoc_plot["label"] = assoc_plot["benchmark"] + " / " + assoc_plot["condition"]
    sns.barplot(data=assoc_plot, y="label", x="point_biserial_r", hue="predictor")
    plt.axvline(0, color="black", lw=0.8, ls="--")
    plt.title("Mechanism Predictors of Item Score Flips")
    plt.xlabel("Point-biserial r with score_flip")
    plt.ylabel("")
    plt.tight_layout()
    plt.savefig(RESULT_DIR / "fig_mechanism_flip_predictors.pdf", bbox_inches="tight")
    plt.show()
```

```python
"""---
## Cell 8b — Temperature Sensitivity Analysis
Asks: does varying inference temperature introduce construct-irrelevant variance?
Uses item-level score flips and descriptive variance components. With N_RUNS=1,
run variance and DIF/IRT-style claims are intentionally omitted.
"""

from scipy.stats import pearsonr

temp_df = analysis_df[analysis_df["cond_type"] == "temp"].copy()

# ── 1. Accuracy vs temperature curve ─────────────────────────────────────────
temp_agg = (temp_df.groupby(["model_pair","benchmark","condition","temperature"])
                   ["correct"].mean().reset_index(name="accuracy"))

fig, axes = plt.subplots(1, len(BENCHMARKS), figsize=(5*len(BENCHMARKS), 4),
                          sharey=False)
for ax, bname in zip(axes, BENCHMARKS):
    sub = temp_agg[temp_agg["benchmark"]==bname]
    for pair, grp in sub.groupby("model_pair"):
        grp_sorted = grp.sort_values("temperature")
        ax.plot(grp_sorted["temperature"], grp_sorted["accuracy"],
                marker="o", label=pair)
    # mark collection temperature
    ax.axvline(COLLECTION_TEMP, color="gray", ls="--", lw=0.8,
                label=f"Collection T={COLLECTION_TEMP}")
    ax.set_xlabel("Temperature"); ax.set_ylabel("Accuracy")
    ax.set_title(bname); ax.legend(fontsize=7)

plt.suptitle("Accuracy vs Decoding Temperature", y=1.02)
plt.tight_layout()
plt.savefig(RESULT_DIR / "fig_temp_accuracy_curve.pdf", bbox_inches="tight")
plt.show()

# ── 2. Pearson r: temperature vs accuracy per benchmark×model ─────────────────
print("\n=== Pearson r: temperature vs accuracy ===")
temp_corr_rows = []
for (pair, bname), grp in temp_agg.groupby(["model_pair","benchmark"]):
    if len(grp) < 3: continue
    r, p = pearsonr(grp["temperature"], grp["accuracy"])
    print(f"  {pair} / {bname}: r={r:.4f}, p={p:.4f}")
    temp_corr_rows.append({"model_pair":pair,"benchmark":bname,"r":r,"p":p})
temp_corr_df = pd.DataFrame(temp_corr_rows)

# ── 3. Descriptive variance components for temperature conditions ────────────
# N_RUNS=1, so run variance is intentionally excluded.
print("\n=== Descriptive Variance: Temperature as Facet ===")
gtemp_rows = []
for (pair, bname), grp in temp_df.groupby(["model_pair","benchmark"]):
    grp = grp.copy()
    item_means   = grp.groupby("item_id")["correct"].mean()
    var_item     = item_means.var()
    temp_means   = grp.groupby("temperature")["correct"].mean()
    var_temp     = temp_means.var()
    ti_means     = grp.groupby(["temperature","item_id"])["correct"].mean()
    var_temp_x_item = ti_means.groupby(level=0).var().mean()
    var_total    = grp["correct"].var()
    var_error    = max(0, var_total - var_item - var_temp - var_temp_x_item)
    n_temp       = grp["temperature"].nunique()
    G_coeff      = var_item / (var_item + var_error / n_temp + 1e-9)
    gtemp_rows.append({
        "model_pair": pair, "benchmark": bname,
        "var_item": var_item, "var_temperature": var_temp,
        "var_temp_x_item": var_temp_x_item,
        "var_error": var_error, "var_total": var_total, "G_coeff": G_coeff,
    })
gtemp_df = pd.DataFrame(gtemp_rows)

vc_cols_t = ["var_item","var_temperature","var_temp_x_item","var_error"]
pct_t = gtemp_df[vc_cols_t].div(gtemp_df["var_total"]+1e-9, axis=0)*100
pct_t.columns = [c+"_%" for c in vc_cols_t]
display_gt = pd.concat([gtemp_df[["benchmark","G_coeff"]], pct_t], axis=1)
print(display_gt.to_string(index=False, float_format="{:.2f}".format))

fig, axes = plt.subplots(1, len(BENCHMARKS), figsize=(5*len(BENCHMARKS), 4))
for ax, bname in zip(axes, BENCHMARKS):
    rows_bname = gtemp_df[gtemp_df["benchmark"]==bname]
    if rows_bname.empty: continue
    row  = rows_bname.iloc[0]
    vals = [row[c] for c in vc_cols_t]
    tot  = sum(vals)+1e-9
    ax.bar([c.replace("var_","") for c in vc_cols_t],
           [v/tot*100 for v in vals],
           color=sns.color_palette("Set2", len(vc_cols_t)))
    ax.set_title(f"{bname}\nG={row['G_coeff']:.3f}")
    ax.set_ylabel("% variance"); ax.tick_params(axis='x', rotation=30)

plt.suptitle("G-Theory: Temperature as Measurement Facet", y=1.02)
plt.tight_layout()
plt.savefig(RESULT_DIR / "fig_temp_gtheory.pdf", bbox_inches="tight")
plt.show()

# ── 4. Per-item temperature sensitivity: flag "thermally unstable" items ──────
# An item is thermally unstable if its correct/incorrect flips across temperatures.
# This is the temperature analogue of DIF.
print("\n=== Thermally Unstable Items (score flips across temperatures) ===")
therm_rows = []
temp_item_scores = (temp_df.groupby(["model_pair","benchmark","item_id","temperature"])
                            ["correct"].mean().reset_index())
for (pair, bname, iid), grp in temp_item_scores.groupby(
        ["model_pair","benchmark","item_id"]):
    scores = grp.sort_values("temperature")["correct"].values
    n_flips = int(np.sum(np.diff(scores.round()) != 0))
    # variance of score across temperatures
    var_score = scores.var()
    therm_rows.append({
        "model_pair": pair, "benchmark": bname, "item_id": iid,
        "n_flips": n_flips, "score_var": var_score,
        "unstable": n_flips >= 2,  # flips at 2+ temperature steps
    })
therm_df = pd.DataFrame(therm_rows)
therm_summary = (therm_df.groupby("benchmark")["unstable"]
                          .agg(n_unstable="sum", n_total="count"))
therm_summary["pct_unstable"] = (therm_summary["n_unstable"]
                                  / therm_summary["n_total"] * 100)
print(therm_summary.to_string(float_format="{:.2f}".format))

# ── 5. Temperature × Speculative interaction ──────────────────────────────────
# Key measurement science question: does temperature sensitivity differ between
# exact speculation (B) and baseline (A)?
# For each item, compute score under A and B, then ask whether the
# T_070 temperature condition gives the same answer as A.
print("\n=== Temperature vs Speculation: score agreement at T=0.7 vs Condition A ===")
agree_rows = []
for (pair, bname), grp in analysis_df.groupby(["model_pair","benchmark"]):
    base_A   = grp[grp["condition"]=="A"].set_index("item_id")["correct"]
    base_T07 = grp[grp["condition"]=="T_070"].set_index("item_id")["correct"]
    base_B   = grp[grp["condition"]=="B"].set_index("item_id")["correct"]
    shared   = base_A.index.intersection(base_T07.index).intersection(base_B.index)
    if len(shared) < 2: continue
    agree_T  = (base_A[shared] == base_T07[shared]).mean()
    agree_B  = (base_A[shared] == base_B[shared]).mean()
    agree_rows.append({
        "model_pair": pair, "benchmark": bname,
        "agree_A_vs_T070": agree_T, "agree_A_vs_B": agree_B,
        "delta": agree_T - agree_B,
    })
agree_df = pd.DataFrame(agree_rows)
print(agree_df.to_string(index=False, float_format="{:.4f}".format))

# Save
temp_corr_df.to_csv(RESULT_DIR / "temp_correlations.csv", index=False)
gtemp_df.to_csv(    RESULT_DIR / "temp_gtheory.csv", index=False)
therm_df.to_csv(    RESULT_DIR / "thermal_instability.csv", index=False)
agree_df.to_csv(    RESULT_DIR / "temp_spec_agreement.csv", index=False)
```

```python
"""---
## Cell 10 — Final Dataset Audit + Helpers
Confirms all downstream analyses use full MMLU/HellaSwag and projected GSM8K.
"""

benchmarks_list = list(analysis_df["benchmark"].unique())
pairs_list = list(analysis_df["model_pair"].unique())

print("Final analysis item counts:")
print(analysis_df.groupby(["benchmark", "analysis_source"])["item_id"]
                 .nunique().to_string())

print("\nRows by benchmark/condition:")
print(analysis_df.groupby(["benchmark", "condition"]).size().unstack(fill_value=0).to_string())

print("\nGSM8K projection note:")
print(f"  Raw GSM8K pilot items: {gsm_obs['item_id'].nunique() if len(gsm_obs) else 0}")
print(f"  Projected GSM8K analysis items: {gsm_projected_df['item_id'].nunique() if len(gsm_projected_df) else 0}")
print("  Raw 20-item GSM8K rows are not included in analysis_df.")

if len(extrap_results):
    rows = []
    for (pair, bname, cond), res in extrap_results.items():
        rows.append({
            "model_pair": pair,
            "benchmark": bname,
            "condition": cond,
            "pilot_mean": res["obs_mean"],
            "projected_ci_lo": res["projected_ci_lo"],
            "projected_ci_hi": res["projected_ci_hi"],
        })
    gsm8k_projection_ci_df = pd.DataFrame(rows)
    print("\nGSM8K projected uncertainty summary:")
    print(gsm8k_projection_ci_df.to_string(index=False, float_format="{:.4f}".format))
    gsm8k_projection_ci_df.to_csv(RESULT_DIR / "gsm8k_projected_ci_summary.csv", index=False)

# ─────────────────────────────────────────────────────────────────────────────
# Helper: run one analysis on a given df, return (figures, tables)
# Each analysis function takes df and a label string, saves nothing itself.
# The comparison cells call each function twice and arrange the plots.
# ─────────────────────────────────────────────────────────────────────────────

def bootstrap_ci_vec(x, n_boot=2000, ci=0.95, seed=SEED):
    rng   = np.random.default_rng(seed)
    boots = [rng.choice(x, size=len(x), replace=True).mean()
             for _ in range(n_boot)]
    lo = np.percentile(boots, 100*(1-ci)/2)
    hi = np.percentile(boots, 100*(1+ci)/2)
    return np.mean(x), lo, hi
```

```python
"""---
## Cell 11 — Analysis 1: Final Aggregate Accuracy
Uses analysis_df only: full MMLU/HellaSwag plus projected GSM8K.
McNemar used instead of t-test because N_RUNS=1.
"""

from statsmodels.stats.contingency_tables import mcnemar as mcnemar_test

def run_analysis1(df_in, label):
    """Returns summary DataFrame and mcnemar DataFrame."""
    cond_order = list(CONDITIONS.keys())
    agg = (df_in.groupby(["model_pair","benchmark","condition","run"])
                ["correct"].mean().reset_index(name="accuracy"))
    summary_rows = []
    for (pair, bname, cond), grp in agg.groupby(["model_pair","benchmark","condition"]):
        accs = grp["accuracy"].values
        mean, lo, hi = bootstrap_ci_vec(accs)
        summary_rows.append({"model_pair": pair, "benchmark": bname,
                               "condition": cond, "mean": mean,
                               "ci_lo": lo, "ci_hi": hi})
    summary = pd.DataFrame(summary_rows)

    # McNemar per-item (works with N_RUNS=1)
    df_r0 = df_in[df_in["run"]==0]
    mcn_rows = []
    for (pair, bname), grp in df_r0.groupby(["model_pair","benchmark"]):
        base = grp[grp["condition"]=="A"].set_index("item_id")["correct"]
        for cond in [c for c in CONDITIONS if c != "A"]:
            comp   = grp[grp["condition"]==cond].set_index("item_id")["correct"]
            shared = base.index.intersection(comp.index)
            if len(shared) < 2: continue
            b_vec  = base[shared].astype(int).values
            c_vec  = comp[shared].astype(int).values
            b_cnt  = int(((b_vec==1)&(c_vec==0)).sum())
            c_cnt  = int(((b_vec==0)&(c_vec==1)).sum())
            if b_cnt + c_cnt == 0:
                p_val = 1.0
            else:
                table  = [[0, b_cnt],[c_cnt, 0]]
                result = mcnemar_test(table, exact=(b_cnt+c_cnt)<25)
                p_val  = result.pvalue
            mcn_rows.append({
                "model_pair": pair, "benchmark": bname, "condition": cond,
                "n_items": len(shared), "delta": c_vec.mean()-b_vec.mean(),
                "b_cnt": b_cnt, "c_cnt": c_cnt, "p": p_val,
            })
    mcn_df = pd.DataFrame(mcn_rows)
    if len(mcn_df):
        mcn_df["p_bonf"] = (mcn_df["p"]*len(mcn_df)).clip(upper=1.0)
        mcn_df["sig"]    = mcn_df["p_bonf"] < 0.05

    return summary, mcn_df

summary_final, mcn_final = run_analysis1(analysis_df, "FINAL ANALYSIS DATASET")
# Backward-compatible aliases used by later summary cells.
summary_pilot = summary_proj = summary_final
mcn_pilot = mcn_proj = mcn_final

# ── Side-by-side plot ─────────────────────────────────────────────
cond_order = list(CONDITIONS.keys())
fig, all_axes = plt.subplots(
    2, len(benchmarks_list),
    figsize=(5*len(benchmarks_list), 8), sharey=False)

for row_i, (summary, label) in enumerate([
        (summary_final, "FINAL: full MMLU/HellaSwag + projected GSM8K"),
        (summary_final, "FINAL: same dataset, repeated for layout compatibility")]):
    for col_j, bname in enumerate(benchmarks_list):
        ax  = all_axes[row_i][col_j]
        sub = summary[summary["benchmark"]==bname]
        for pair, grp in sub.groupby("model_pair"):
            grp = grp.set_index("condition").reindex(cond_order)
            ax.errorbar(range(len(cond_order)), grp["mean"],
                        yerr=[grp["mean"]-grp["ci_lo"], grp["ci_hi"]-grp["mean"]],
                        marker="o", label=pair, capsize=4)
        ax.set_title(f"{bname}\n{label}", fontsize=9)
        ax.set_xticks(range(len(cond_order)))
        ax.set_xticklabels(cond_order, rotation=30, ha="right")
        ax.set_ylabel("Accuracy")
        ax.legend(fontsize=7)
        if row_i == 1:
            ax.set_facecolor("#fff8f0")  # tint projected panels

plt.suptitle("Analysis 1: Accuracy by Condition\nFinal dataset = full MMLU/HellaSwag + projected GSM8K",
              y=1.01, fontsize=11)
plt.tight_layout()
plt.savefig(RESULT_DIR / "fig1_accuracy_comparison.pdf", bbox_inches="tight")
plt.show()

print("\n=== McNemar — FINAL ANALYSIS DATASET ===")
print(mcn_final.to_string(index=False, float_format="{:.4f}".format))

summary_final.to_csv(RESULT_DIR / "a1_summary_final.csv", index=False)
mcn_final.to_csv(RESULT_DIR / "a1_mcnemar_final.csv", index=False)
```

```python
"""---
## Cell 12 — Analysis 2: G-Theory Variance Components (Final Dataset)
"""

def run_gtheory(df_in):
    """Descriptive variance decomposition over items and decoding conditions.

    N_RUNS=1, so run variance is intentionally excluded. This is not a full
    crossed G-study; it is a measurement-facet decomposition for the observed
    cached traces.
    """
    rows = []
    for (pair, bname), grp in df_in.groupby(["model_pair","benchmark"]):
        grp = grp.copy()
        item_means      = grp.groupby("item_id")["correct"].mean()
        var_item        = item_means.var()
        cond_means      = grp.groupby("condition")["correct"].mean()
        var_condition   = cond_means.var()
        ci_means        = grp.groupby(["condition","item_id"])["correct"].mean()
        var_cond_x_item = ci_means.groupby(level=0).var().mean()
        var_total       = grp["correct"].var()
        var_error       = max(0, var_total - var_item - var_condition
                               - var_cond_x_item)
        n_cond  = grp["condition"].nunique()
        G_coeff = var_item / (var_item + var_error/n_cond + 1e-9)
        rows.append({"model_pair": pair, "benchmark": bname,
                      "var_item": var_item, "var_condition": var_condition,
                      "var_cond_x_item": var_cond_x_item,
                      "var_error": var_error, "var_total": var_total,
                      "G_coeff": G_coeff})
    return pd.DataFrame(rows)

gt_final = run_gtheory(analysis_df)
gt_pilot = gt_proj = gt_final

vc_cols = ["var_item","var_condition","var_cond_x_item","var_error"]

fig, all_axes = plt.subplots(
    2, len(benchmarks_list),
    figsize=(5*len(benchmarks_list), 8))

for row_i, (gt, label) in enumerate([
        (gt_final, "FINAL analysis dataset"),
        (gt_final, "FINAL analysis dataset, repeated for layout compatibility")]):
    for col_j, bname in enumerate(benchmarks_list):
        ax   = all_axes[row_i][col_j]
        rows = gt[gt["benchmark"]==bname]
        if rows.empty: ax.set_title(f"{bname}\nno data"); continue
        row  = rows.iloc[0]
        vals  = [row[c] for c in vc_cols]
        total = sum(vals) + 1e-9
        ax.bar([c.replace("var_","") for c in vc_cols],
               [v/total*100 for v in vals],
               color=sns.color_palette("Set2", len(vc_cols)))
        ax.set_title(f"{bname}\n{label}\nG={row['G_coeff']:.3f}", fontsize=9)
        ax.set_ylabel("% variance")
        ax.tick_params(axis='x', rotation=30)
        if row_i == 1: ax.set_facecolor("#fff8f0")

plt.suptitle("Analysis 2: G-Theory Variance Components\nFinal dataset only",
              y=1.01, fontsize=11)
plt.tight_layout()
plt.savefig(RESULT_DIR / "fig2_gtheory_comparison.pdf", bbox_inches="tight")
plt.show()

gt_final.to_csv(RESULT_DIR / "a2_gtheory_final.csv", index=False)
```

```python
"""---
## Cell 13 — Removed Analysis: IRT + DIF (Not Appropriate for N_RUNS=1)
"""

def irt_2pl_prob(theta, a, b):
    return 1.0 / (1.0 + np.exp(-a * (theta - b)))

def irt_neg_log_likelihood(params, R):
    N, I = R.shape
    thetas = params[:N]
    a      = np.exp(params[N:N+I])
    b      = params[N+I:]
    P   = irt_2pl_prob(thetas[:,None], a[None,:], b[None,:])
    P   = np.clip(P, 1e-6, 1-1e-6)
    nll = -np.sum(R*np.log(P) + (1-R)*np.log(1-P))
    nll += 0.1*np.sum(thetas**2) + 0.1*np.sum(b**2)
    return nll

def fit_irt(R, n_restarts=3):
    N, I = R.shape
    best_res = None
    rng = np.random.default_rng(SEED)
    for _ in range(n_restarts):
        x0  = np.concatenate([rng.normal(0,0.5,N), rng.normal(0,0.3,I), rng.normal(0,0.5,I)])
        res = minimize(irt_neg_log_likelihood, x0, args=(R,),
                       method="L-BFGS-B", options={"maxiter":2000,"ftol":1e-8})
        if best_res is None or res.fun < best_res.fun: best_res = res
    thetas = best_res.x[:N]
    a      = np.exp(best_res.x[N:N+I])
    b      = best_res.x[N+I:]
    return thetas, a, b, -best_res.fun

def build_response_matrix(df_sub, model_col, item_col, correct_col):
    pivot = df_sub.pivot_table(index=model_col, columns=item_col,
                                values=correct_col, aggfunc="mean")
    return (pivot.fillna(0.5).values > 0.5).astype(float), \
           list(pivot.index), list(pivot.columns)

def run_dif(df_in, label):
    df_mean = (df_in.groupby(["model_pair","benchmark","condition","item_id"])
                     ["correct"].mean().reset_index())
    dif_rows = []
    for (pair, bname), grp in df_mean.groupby(["model_pair","benchmark"]):
        Ra, models_a, items_a = build_response_matrix(
            grp[grp["condition"]=="A"], "model_pair","item_id","correct")
        Rb, models_b, items_b = build_response_matrix(
            grp[grp["condition"]=="B"], "model_pair","item_id","correct")
        if Ra.shape[0] < 3:
            print(f"  {label}/{bname}: skipped (need ≥3 models for IRT)")
            continue
        common = [i for i in items_a if i in items_b]
        ia = [items_a.index(i) for i in common]
        ib = [items_b.index(i) for i in common]
        if len(common) < 5:
            print(f"  {label}/{bname}: skipped (<5 common items)")
            continue
        print(f"  Fitting IRT {label}/{bname} ({len(common)} items)...", end=" ")
        th_a, a_a, b_a, _ = fit_irt(Ra[:,ia])
        th_b, a_b, b_b, _ = fit_irt(Rb[:,ib])
        for k, iid in enumerate(common):
            delta_b   = b_b[k] - b_a[k]; delta_a = a_b[k] - a_a[k]
            chi2_stat = (delta_b/0.3)**2 + (delta_a/0.3)**2
            p_dif     = 1 - chi2.cdf(chi2_stat, df=2)
            dif_rows.append({"model_pair":pair,"benchmark":bname,"item_id":iid,
                              "b_A":b_a[k],"b_B":b_b[k],"delta_b":delta_b,
                              "a_A":a_a[k],"a_B":a_b[k],"delta_a":delta_a,
                              "chi2":chi2_stat,"p_dif":p_dif})
        print("done")
    dif_df = pd.DataFrame(dif_rows)
    if len(dif_df):
        _, p_adj, _, _ = multipletests(dif_df["p_dif"], method="fdr_bh")
        dif_df["p_fdr"] = p_adj; dif_df["DIF"] = dif_df["p_fdr"] < 0.05
    return dif_df

print("IRT/DIF analysis skipped: not appropriate with N_RUNS=1 and few model rows.")
dif_final = pd.DataFrame()
dif_pilot = dif_proj = dif_final

def plot_dif(dif_df, axes_row, label):
    for col_j, bname in enumerate(benchmarks_list):
        ax  = axes_row[col_j]
        sub = dif_df[dif_df["benchmark"]==bname] if len(dif_df) else pd.DataFrame()
        if sub.empty: ax.set_title(f"{bname}\n{label}\n(no IRT data)"); continue
        dif_items     = sub[sub["DIF"]==True]
        non_dif_items = sub[sub["DIF"]==False]
        ax.scatter(non_dif_items["b_A"], non_dif_items["b_B"],
                    alpha=0.4, s=10, color="steelblue", label="Non-DIF")
        ax.scatter(dif_items["b_A"], dif_items["b_B"],
                    alpha=0.8, s=20, color="tomato",
                    label=f"DIF ({len(dif_items)})")
        lim = [sub[["b_A","b_B"]].min().min()-0.5,
               sub[["b_A","b_B"]].max().max()+0.5]
        ax.plot(lim, lim, "k--", lw=0.8, alpha=0.5)
        ax.set_xlim(lim); ax.set_ylim(lim)
        ax.set_xlabel("b (Cond A)"); ax.set_ylabel("b (Cond B)")
        ax.set_title(f"{bname}\n{label}", fontsize=9)
        ax.legend(fontsize=7)

# No IRT/DIF figures are produced for the final analysis.
dif_final.to_csv(RESULT_DIR / "a3_dif_skipped.csv", index=False)
```

```python
"""---
## Cell 14 — Analysis 4: Construct-Level Stability (Final Dataset)
"""

def run_construct_stability(df_in):
    cs = (df_in.groupby(["model_pair","benchmark","condition","construct"])
               ["correct"].mean().reset_index(name="accuracy"))
    ref = (cs[cs["condition"]=="A"]
             .rename(columns={"accuracy":"acc_A"})
             [["model_pair","benchmark","construct","acc_A"]])
    merged = cs.merge(ref, on=["model_pair","benchmark","construct"])
    merged["delta_from_A"] = merged["accuracy"] - merged["acc_A"]
    pivot = (merged[merged["condition"]!="A"]
             .groupby(["construct","condition"])["delta_from_A"].mean()
             .unstack("condition"))
    return merged, pivot

cs_merged_final, cs_pivot_final = run_construct_stability(analysis_df)
cs_merged_pilot = cs_merged_proj = cs_merged_final
cs_pivot_pilot = cs_pivot_proj = cs_pivot_final

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, (pivot, label) in zip(axes, [
        (cs_pivot_final, "FINAL analysis dataset"),
        (cs_pivot_final, "FINAL analysis dataset, repeated for layout compatibility")]):
    if pivot.empty: ax.set_title(f"{label}\n(no data)"); continue
    sns.heatmap(pivot, annot=True, fmt=".3f", center=0,
                cmap="RdBu_r", ax=ax, linewidths=0.5)
    ax.set_title(f"Accuracy Δ vs Baseline A\n{label}", fontsize=10)
    ax.set_xlabel("Condition"); ax.set_ylabel("Construct")

plt.suptitle("Analysis 4: Construct-Level Stability\nFinal dataset only",
              y=1.01, fontsize=11)
plt.tight_layout()
plt.savefig(RESULT_DIR / "fig4_construct_comparison.pdf", bbox_inches="tight")
plt.show()

cs_merged_final.to_csv(RESULT_DIR / "a4_construct_final.csv", index=False)
```

```python
"""---
## Cell 14 — Analysis 6: Acceptance Rate Regression + Draft Quality Moderator
Extended from original: adds accuracy_gap as moderator and interaction term.
Research question: does draft model quality moderate score degradation under
approximate speculation, and does this vary by construct?

With N_RUNS=1 these regressions are exploratory descriptive summaries over
benchmark/condition cells. Report slopes/effect sizes, not confirmatory p-values.
"""

print("Exploratory regression note: N_RUNS=1, so slopes are descriptive rather than confirmatory.")

import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

approx_conds = [c for c in SPEC_CONDITIONS if c.startswith("C_")]

# Mean accept_rate and accuracy per group
ar_df = (analysis_df[analysis_df["condition"].isin(approx_conds)]
           .groupby(["model_pair","benchmark","construct","condition","run"])
           .agg(accuracy=("correct","mean"),
                mean_accept_rate=("accept_rate","mean"))
           .reset_index())

# Merge baseline accuracy
base_acc = (analysis_df[analysis_df["condition"]=="A"]
              .groupby(["model_pair","benchmark","run"])
              ["correct"].mean().reset_index(name="acc_A"))
ar_df = ar_df.merge(base_acc, on=["model_pair","benchmark","run"])
ar_df["delta_acc"]      = ar_df["accuracy"] - ar_df["acc_A"]
ar_df["rejection_rate"] = 1 - ar_df["mean_accept_rate"]

# Merge accuracy gap (the moderator)
gap_mean = (acc_gap_df.groupby(["model_pair","benchmark"])
                      ["accuracy_gap"].mean().reset_index())
ar_df = ar_df.merge(gap_mean, on=["model_pair","benchmark"], how="left")
ar_df["accuracy_gap"] = ar_df["accuracy_gap"].fillna(0.0)

# Interaction term (mean-centered to reduce collinearity)
ar_df["rejection_rate_c"] = (ar_df["rejection_rate"]
                               - ar_df["rejection_rate"].mean())
ar_df["accuracy_gap_c"]   = (ar_df["accuracy_gap"]
                               - ar_df["accuracy_gap"].mean())
ar_df["interaction"]      = (ar_df["rejection_rate_c"]
                               * ar_df["accuracy_gap_c"])

# ── Model 1 (original): delta_acc ~ rejection_rate ────────────────────────────
print("\n=== Model 1 (original): delta_acc ~ rejection_rate ===")
reg1_rows = []
for construct, grp in ar_df.groupby("construct"):
    if len(grp) < 5: continue
    X   = sm.add_constant(grp["rejection_rate"])
    res = sm.OLS(grp["delta_acc"], X).fit()
    b1  = res.params.get("rejection_rate", np.nan)
    p1  = res.pvalues.get("rejection_rate", np.nan)
    print(f"  {construct}: β₁={b1:.4f}, p={p1:.4f}, R²={res.rsquared:.3f}")
    reg1_rows.append({"construct":construct,"beta_rejection":b1,
                       "p_rejection":p1,"r2_model1":res.rsquared})
reg1_df = pd.DataFrame(reg1_rows)

# ── Model 2 (extended): delta_acc ~ rejection_rate + accuracy_gap ─────────────
print("\n=== Model 2 (additive): delta_acc ~ rejection_rate + accuracy_gap ===")
reg2_rows = []
for construct, grp in ar_df.groupby("construct"):
    if len(grp) < 5: continue
    X   = sm.add_constant(grp[["rejection_rate_c","accuracy_gap_c"]])
    res = sm.OLS(grp["delta_acc"], X).fit()
    b_rej = res.params.get("rejection_rate_c", np.nan)
    b_gap = res.params.get("accuracy_gap_c",   np.nan)
    p_rej = res.pvalues.get("rejection_rate_c", np.nan)
    p_gap = res.pvalues.get("accuracy_gap_c",   np.nan)
    print(f"  {construct}: β_rejection={b_rej:.4f}(p={p_rej:.3f})  "
          f"β_gap={b_gap:.4f}(p={p_gap:.3f})  R²={res.rsquared:.3f}")
    reg2_rows.append({"construct":construct,
                       "beta_rejection":b_rej,"p_rejection":p_rej,
                       "beta_gap":b_gap,"p_gap":p_gap,
                       "r2_model2":res.rsquared})
reg2_df = pd.DataFrame(reg2_rows)

# ── Model 3 (moderated): delta_acc ~ rejection_rate * accuracy_gap ────────────
print("\n=== Model 3 (moderated): delta_acc ~ rejection_rate × accuracy_gap ===")
print("    Interaction β significant → draft quality moderates score degradation")
reg3_rows = []
for construct, grp in ar_df.groupby("construct"):
    if len(grp) < 5: continue
    X   = sm.add_constant(
            grp[["rejection_rate_c","accuracy_gap_c","interaction"]])
    res = sm.OLS(grp["delta_acc"], X).fit()
    b_rej  = res.params.get("rejection_rate_c", np.nan)
    b_gap  = res.params.get("accuracy_gap_c",   np.nan)
    b_int  = res.params.get("interaction",       np.nan)
    p_rej  = res.pvalues.get("rejection_rate_c", np.nan)
    p_gap  = res.pvalues.get("accuracy_gap_c",   np.nan)
    p_int  = res.pvalues.get("interaction",       np.nan)
    # R² change from Model 2 → Model 3
    r2_m2  = reg2_df[reg2_df["construct"]==construct]["r2_model2"].values
    r2_change = res.rsquared - (r2_m2[0] if len(r2_m2) else 0)
    print(f"  {construct}:")
    print(f"    β_rejection={b_rej:.4f}(p={p_rej:.3f})  "
          f"β_gap={b_gap:.4f}(p={p_gap:.3f})  "
          f"β_interaction={b_int:.4f}(p={p_int:.3f})")
    print(f"    R²={res.rsquared:.3f}  ΔR²(interaction)={r2_change:+.3f}")
    reg3_rows.append({"construct":construct,
                       "beta_rejection":b_rej,"p_rejection":p_rej,
                       "beta_gap":b_gap,"p_gap":p_gap,
                       "beta_interaction":b_int,"p_interaction":p_int,
                       "r2_model3":res.rsquared,"delta_r2":r2_change})
reg3_df = pd.DataFrame(reg3_rows)

# ── Plot 1: original scatter (Model 1) ───────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
palette = sns.color_palette("tab10", ar_df["construct"].nunique())
ax = axes[0]
for (construct, grp), color in zip(ar_df.groupby("construct"), palette):
    ax.scatter(grp["rejection_rate"], grp["delta_acc"],
                alpha=0.4, s=20, color=color, label=construct)
    row = reg1_df[reg1_df["construct"]==construct]
    if not row.empty:
        xr = np.linspace(grp["rejection_rate"].min(),
                          grp["rejection_rate"].max(), 50)
        b1 = row.iloc[0]["beta_rejection"]
        b0 = grp["delta_acc"].mean() - b1*grp["rejection_rate"].mean()
        ax.plot(xr, b0+b1*xr, color=color, lw=2)
ax.axhline(0, color="black", lw=0.8, ls="--")
ax.set_xlabel("Rejection Rate (1−α̅)")
ax.set_ylabel("Accuracy Δ from Baseline")
ax.set_title("Model 1: Score Degradation vs Rejection Rate")
ax.legend(title="Construct", fontsize=8)

# ── Plot 2: moderation plot — one line per accuracy_gap quartile ──────────────
ax = axes[1]
ar_df["gap_quartile"] = pd.qcut(ar_df["accuracy_gap"], q=2,
                                  labels=["Small gap\n(similar models)",
                                          "Large gap\n(very different models)"])
gap_palette = sns.color_palette("coolwarm", 2)
for (construct, gap_q), grp in ar_df.groupby(["construct","gap_quartile"]):
    color = gap_palette[0] if "Small" in str(gap_q) else gap_palette[1]
    ls    = "-" if list(ar_df["construct"].unique()).index(construct) % 2 == 0 else "--"
    if len(grp) < 3: continue
    xr = np.linspace(grp["rejection_rate"].min(),
                      grp["rejection_rate"].max(), 50)
    try:
        X   = sm.add_constant(grp["rejection_rate"])
        res = sm.OLS(grp["delta_acc"], X).fit()
        b1  = res.params.get("rejection_rate", 0)
        b0  = grp["delta_acc"].mean() - b1*grp["rejection_rate"].mean()
        ax.plot(xr, b0+b1*xr, color=color, lw=2, ls=ls,
                label=f"{construct} / {gap_q}")
    except Exception:
        pass
ax.axhline(0, color="black", lw=0.8, ls="--")
ax.set_xlabel("Rejection Rate (1−α̅)")
ax.set_ylabel("Accuracy Δ from Baseline")
ax.set_title("Moderation Plot: Draft Quality × Rejection Rate\n"
              "Diverging slopes = significant interaction")
ax.legend(fontsize=7, loc="lower left")

plt.tight_layout()
plt.savefig(RESULT_DIR / "fig6_acceptance_moderation.pdf", bbox_inches="tight")
plt.show()

# ── Plot 3: draft accuracy gap vs beta_rejection per construct ────────────────
# If draft quality moderates degradation, constructs with larger gaps
# should have steeper (more negative) slopes.
if len(reg3_df) > 1:
    fig, ax = plt.subplots(figsize=(7, 5))
    gap_by_construct = (ar_df.groupby("construct")["accuracy_gap"]
                              .mean().reset_index())
    plot_data = reg3_df.merge(gap_by_construct, on="construct")
    ax.scatter(plot_data["accuracy_gap"], plot_data["beta_rejection"],
                s=80, zorder=5)
    for _, row_p in plot_data.iterrows():
        ax.annotate(row_p["construct"],
                    (row_p["accuracy_gap"], row_p["beta_rejection"]),
                    textcoords="offset points", xytext=(6,4), fontsize=9)
    # fit line
    if len(plot_data) >= 3:
        xf = np.linspace(plot_data["accuracy_gap"].min(),
                          plot_data["accuracy_gap"].max(), 50)
        try:
            coefs = np.polyfit(plot_data["accuracy_gap"],
                                plot_data["beta_rejection"], 1)
            ax.plot(xf, np.polyval(coefs, xf), "k--", lw=1.5, alpha=0.7)
        except Exception:
            pass
    ax.axhline(0, color="gray", lw=0.8, ls=":")
    ax.set_xlabel("Mean Draft-Target Accuracy Gap")
    ax.set_ylabel("β₁ (slope of score degradation)")
    ax.set_title("Does Draft Quality Predict Degradation Slope?\n"
                  "Negative slope → larger gap = steeper degradation")
    plt.tight_layout()
    plt.savefig(RESULT_DIR / "fig6b_gap_vs_slope.pdf", bbox_inches="tight")
    plt.show()

# Save
reg1_df.to_csv(RESULT_DIR / "reg_model1.csv", index=False)
reg2_df.to_csv(RESULT_DIR / "reg_model2.csv", index=False)
reg3_df.to_csv(RESULT_DIR / "reg_model3.csv", index=False)
ar_df.to_csv(  RESULT_DIR / "acceptance_rate_data.csv", index=False)

print("\n=== R² improvement: Model 1 → Model 3 (moderation adds explained variance) ===")
r2_compare = reg1_df[["construct","r2_model1"]].merge(
    reg3_df[["construct","r2_model3","delta_r2","p_interaction"]],
    on="construct")
print(r2_compare.to_string(index=False, float_format="{:.4f}".format))
```

```python
"""---
## Cell 15 — Analysis 6: Acceptance Rate Regression  (Pilot vs Projected)
"""

print("="*70)
print("HYPOTHESIS EVALUATION SUMMARY")
print("="*70)

pvals = mcn_pilot
if "p_bonf" not in pvals.columns and "p" in pvals.columns:
    pvals = pvals.assign(p_bonf=(pvals["p"] * len(pvals)).clip(upper=1.0))
dif_df = dif_pilot
# Kendall-tau analysis is optional; keep the summary runnable if it is absent.
tau_df = globals().get("tau_pilot", pd.DataFrame())

# H1: Aggregate scores unchanged under exact speculation (Cond B)
h1 = pvals[pvals["condition"]=="B"][["benchmark","delta","p_bonf","sig"]]
print("\nH1 — Aggregate accuracy: Condition B vs A")
print(h1.to_string(index=False))

# Core RQ: Does speculative decoding become a measurement facet?
print("\nCore RQ — Speculative decoding as a measurement facet")
if len(spec_curve_df):
    rq = (spec_curve_df.groupby(["benchmark", "condition"])
                       .agg(mean_delta=("accuracy_delta_vs_A", "mean"),
                            mean_flip_rate=("score_flip_rate", "mean"),
                            mean_accept_rate=("mean_accept_rate", "mean"))
                       .reset_index())
    print(rq.to_string(index=False, float_format="{:.4f}".format))
if len(temp_positive_control_df):
    print("\n  Positive control comparison: speculation vs temperature")
    print(temp_positive_control_df.to_string(index=False,
                                             float_format="{:.4f}".format))
if len(mechanism_assoc_df):
    print("\n  Mechanism evidence: predictors associated with score flips")
    print(mechanism_assoc_df.sort_values("p").head(10)
                            .to_string(index=False, float_format="{:.4f}".format))

# H2: Item instability replaces DIF/IRT for this N_RUNS=1 design.
print("\nH2 — Item instability under decoding perturbations")
if len(score_flip_df):
    h2 = (score_flip_df.groupby(["benchmark", "condition"])
                       .agg(score_flip_rate=("score_flip", "mean"),
                            token_change_rate=("token_changed", "mean"),
                            n_items=("item_id", "nunique"))
                       .reset_index())
    print(h2.to_string(index=False, float_format="{:.4f}".format))
else:
    print("  No score-flip rows available.")

# H3: Construct sensitivity differs across math, knowledge, and commonsense
if len(reg1_df) > 0:
    print("\nH3 — β_rejection by construct")
    print(reg1_df[["construct", "beta_rejection", "p_rejection", "r2_model1"]]
          .to_string(index=False, float_format="{:.4f}".format))

# H4 (NEW): Temperature introduces construct-irrelevant variance
print("\n H4 (NEW) — Temperature sensitivity")
if len(temp_corr_df) > 0:
    print("  Pearson r (temperature vs accuracy):")
    print(temp_corr_df.to_string(index=False, float_format="{:.4f}".format))
if len(therm_summary) > 0:
    print("  Thermally unstable items (score flips ≥2 temperature steps):")
    print(therm_summary.to_string(float_format="{:.2f}".format))
    print(f"  H4 supported (>5% unstable items): "
          f"{(therm_summary['pct_unstable'] > 5).any()}")
if len(gtemp_df) > 0:
    print("  % variance explained by temperature facet:")
    print(gtemp_df[["benchmark","var_temperature","var_total"]]
          .assign(pct=lambda x: x["var_temperature"]/x["var_total"]*100)
          [["benchmark","pct"]].to_string(index=False, float_format="{:.2f}".format))

# H5 (NEW): Draft quality moderation, exploratory only with N_RUNS=1
print("\n H5 (NEW) — Draft quality moderation (exploratory)")
if len(reg3_df) > 0:
    print("  Interaction slopes and R² changes are descriptive; do not treat p-values as confirmatory.")
    print(reg3_df[["construct", "beta_interaction", "delta_r2", "p_interaction"]]
          .to_string(index=False, float_format="{:.4f}".format))
    print("\n  Model comparison (R² Model1 → Model3):")
    print(r2_compare.to_string(index=False, float_format="{:.4f}".format))

# H6: Ranking stability (original H5, renumbered)
if len(tau_df) > 0:
    print("\nH6 — Kendall τ vs Baseline A")
    print(tau_df[["condition","kendall_tau","rank_inversions"]].to_string(index=False))
    low_tau = tau_df[tau_df["kendall_tau"] < 0.90]
    print(f"  H6 supported for conditions: {list(low_tau['condition'])}")

print("\n" + "="*70)
print("Output files in:", RESULT_DIR.resolve())
for f in sorted(RESULT_DIR.iterdir()):
    print(f"  {f.name}")
```
